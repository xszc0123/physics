```python
#등속원운동하는 물체의 주기, 속력, 가속도 구하기
```

```python
m = 0.15
r = 0.60
f = 2.0
pi = 3.14
```

```python
# (a) 주기는 얼마인가?
T = 1 / f
print("(a) 주기는 얼마인가? : ", T, "s")
```

```python
# (b) 속력은 얼마인가?
v = 2 * pi * r / T
print("(b) 속력은 얼마인가? : ", v, "m/s")
```

```python
# (c) 구심가속도는 얼마인가?
ac = v * v / r
print("(c) 구심가속도는 얼마인가? : ", ac, "m/s*s")
```
