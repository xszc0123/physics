```python
r = 500
dt = 10.5
dv = 32
```

```python
# (a) 접선 가속도는 얼마인가?
at = dv / dt
print("(a) 접선 가속도는 얼마인가? : {:.2f}m/s*s".format(at))
```

```python
# (b) 속력이 35m/s일 때 지름 방향의 가속도는 얼마인가?
v = 35
ar = v * v / r
print("(b) 속력이 35m/s일 때 지름 방향의 가속도는 얼마인가? : {:.2f}m/s*s".format(ar))
```
